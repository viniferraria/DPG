```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

import pandas as pd


from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

def plot_pca(df, title):
    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(df.drop(columns=["Y"]))
    
    plt.figure(figsize=(8, 6))
    plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df["Y"], cmap='viridis', edgecolor='k')
    plt.title(title)
    plt.xlabel(f'Principal Component 1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
    plt.ylabel(f'Principal Component 2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
    plt.colorbar(label='Class Label')
    plt.grid()
    plt.show()

df = pd.read_csv("../test_datasets/scenario_1.csv",)
df2_v1 = pd.read_csv("../test_datasets/scenario_2_updated.csv",)
df3 = pd.read_csv("../test_datasets/scenario_3.csv",)
df5 = pd.read_csv("../test_datasets/scenario_5.csv",)

plot_pca(df, "PCA of Scenario 1")
plot_pca(df2_v1, "PCA of Scenario 2 Updated")
plot_pca(df3, "PCA of Scenario 3")
plot_pca(df5, "PCA of Scenario 5")
```
